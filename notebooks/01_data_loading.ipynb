# 01. Data Loading and Initial Exploration

This notebook loads the India Job Market & Salary Dataset and performs initial exploration.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings
warnings.filterwarnings('ignore')

# Set display options
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

# Set plotting style
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')
```

## Define Paths

```python
# Define paths
ROOT_DIR = os.path.dirname(os.path.dirname(os.path.abspath('__file__')))
DATA_DIR = os.path.join(ROOT_DIR, 'data', 'raw')
PROCESSED_DIR = os.path.join(ROOT_DIR, 'data', 'processed')
RESULTS_DIR = os.path.join(ROOT_DIR, 'results')

print(f"Root Directory: {ROOT_DIR}")
print(f"Data Directory: {DATA_DIR}")
print(f"Processed Data Directory: {PROCESSED_DIR}")
print(f"Results Directory: {RESULTS_DIR}")
```

## Load the Dataset

```python
# List files in data directory
print("Files in data directory:")
for f in os.listdir(DATA_DIR):
    print(f"  - {f}")
```

```python
# Load the dataset - Auto-detect file
import glob

csv_files = glob.glob(os.path.join(DATA_DIR, '*.csv'))
xlsx_files = glob.glob(os.path.join(DATA_DIR, '*.xlsx'))
all_files = csv_files + xlsx_files

if all_files:
    file_path = all_files[0]
    print(f"Found file: {os.path.basename(file_path)}")
    
    if file_path.endswith('.csv'):
        df = pd.read_csv(file_path)
    else:
        df = pd.read_excel(file_path)
    
    print(f"Dataset loaded successfully!")
    print(f"Shape: {df.shape}")
else:
    print("No data files found in data/raw/")
    print("\nPlease download the dataset from Kaggle:")
    print("https://www.kaggle.com/datasets/kapoorprakhar/india-job-market-and-salary-dataset")
    print("\nOr try these alternative datasets:")
    print("- https://www.kaggle.com/datasets/ankushnarwade/indian-fresher-salary-and-skill-trends-2025")
    print("- https://www.kaggle.com/datasets/saurabhbadole/latest-data-science-job-salaries-2024")
```

## Auto-Detect Key Columns

```python
# Auto-detect column mappings
COLUMN_MAPPING = {}

for col in df.columns:
    col_lower = col.lower().strip()
    
    # Salary column detection
    if 'salary' in col_lower and 'income' not in col_lower:
        COLUMN_MAPPING['salary'] = col
    
    # Job title detection
    elif 'job' in col_lower and 'title' in col_lower:
        COLUMN_MAPPING['job_title'] = col
    
    # Company detection
    elif 'company' in col_lower and 'name' in col_lower:
        COLUMN_MAPPING['company'] = col
    
    # Location detection
    elif 'location' in col_lower or 'city' in col_lower:
        COLUMN_MAPPING['location'] = col
    
    # Experience detection
    elif 'experience' in col_lower or 'years' in col_lower:
        COLUMN_MAPPING['experience'] = col

print("Detected Column Mappings:")
for key, col in COLUMN_MAPPING.items():
    print(f"  {key}: '{col}'")
```

```python
# Missing values
print("Missing Values:")
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct})
missing_df = missing_df[missing_df['Missing Count'] > 0]
if len(missing_df) > 0:
    print(missing_df)
else:
    print("No missing values found!")
```

## Column Overview

```python
# Display column descriptions
for col in df.columns:
    print(f"\n{col}:")
    print(f"  Unique values: {df[col].nunique()}")
    if df[col].dtype == 'object':
        print(f"  Sample values: {df[col].unique()[:5]}")
```

## Save Processed Data

```python
# Create processed directory if not exists
os.makedirs(PROCESSED_DIR, exist_ok=True)

# Save a copy for further analysis
df.to_csv(os.path.join(PROCESSED_DIR, 'raw_data.csv'), index=False)

# Save column mapping for other notebooks
import json
with open(os.path.join(PROCESSED_DIR, 'column_mapping.json'), 'w') as f:
    json.dump(COLUMN_MAPPING, f, indent=2)

print("Raw data and column mapping saved!")
```
