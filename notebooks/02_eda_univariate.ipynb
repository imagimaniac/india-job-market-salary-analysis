# 02. Univariate Analysis

This notebook performs univariate analysis on the India Job Market & Salary Dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings
warnings.filterwarnings('ignore')

# Set paths
ROOT_DIR = os.path.dirname(os.path.dirname(os.path.abspath('__file__')))
PROCESSED_DIR = os.path.join(ROOT_DIR, 'data', 'processed')
RESULTS_DIR = os.path.join(ROOT_DIR, 'results')

# Load data
df = pd.read_csv(os.path.join(PROCESSED_DIR, 'raw_data.csv'))
print(f"Loaded data with shape: {df.shape}")
```

## Salary Distribution Analysis

```python
# Identify salary column(s)
salary_cols = [col for col in df.columns if 'salary' in col.lower()]
print(f"Salary columns found: {salary_cols}")
```

```python
# Salary column - update based on actual column name
salary_col = salary_cols[0] if salary_cols else 'salary'  # Update this

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram
axes[0].hist(df[salary_col], bins=30, edgecolor='black', alpha=0.7)
axes[0].set_xlabel('Salary (INR)')
axes[0].set_ylabel('Frequency')
axes[0].set_title('Salary Distribution - Histogram')
axes[0].axvline(df[salary_col].mean(), color='red', linestyle='--', label=f"Mean: {df[salary_col].mean():,.0f}")
axes[0].axvline(df[salary_col].median(), color='green', linestyle='--', label=f"Median: {df[salary_col].median():,.0f}")
axes[0].legend()

# Boxplot
axes[1].boxplot(df[salary_col].dropna())
axes[1].set_ylabel('Salary (INR)')
axes[1].set_title('Salary Distribution - Boxplot')

plt.tight_layout()
plt.savefig(os.path.join(RESULTS_DIR, 'figures', 'salary_distribution.png'), dpi=150, bbox_inches='tight')
plt.show()

print(f"\nSalary Statistics:")
print(df[salary_col].describe())
```

## Categorical Variables Analysis

```python
# Identify categorical columns
cat_cols = df.select_dtypes(include=['object']).columns.tolist()
print(f"Categorical columns: {cat_cols}")
```

```python
# Plot top 10 for each categorical column
for col in cat_cols[:5]:  # Analyze first 5 categorical columns
    plt.figure(figsize=(10, 5))
    top_10 = df[col].value_counts().head(10)
    # Check if column has many unique values
    if df[col].nunique() > 10:
        sns.barplot(x=top_10.values, y=top_10.index, palette='viridis')
    else:
        sns.barplot(x=top_10.values, y=top_10.index, palette='viridis')

    plt.xlabel('Count')
    plt.ylabel(col)
    plt.title(f'Top 10 {col}')
    plt.tight_layout()
    plt.savefig(os.path.join(RESULTS_DIR, 'figures', f'top10_{col}.png'), dpi=150, bbox_inches='tight')
    plt.show()
```

## Summary Statistics Table

```python
# Create summary statistics
summary_stats = df.describe(include='all').T
summary_stats.to_csv(os.path.join(RESULTS_DIR, 'tables', 'summary_statistics.csv'))
print("Summary statistics saved!")
summary_stats
```
