# 03. Bivariate & Multivariate Analysis

Analyze relationships between variables.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings
warnings.filterwarnings('ignore')

ROOT_DIR = os.path.dirname(os.path.dirname(os.path.abspath('__file__')))
PROCESSED_DIR = os.path.join(ROOT_DIR, 'data', 'processed')
RESULTS_DIR = os.path.join(ROOT_DIR, 'results')

df = pd.read_csv(os.path.join(PROCESSED_DIR, 'raw_data.csv'))
```

## Salary by Job Title

```python
# Salary by Job Title
# Update column names based on your dataset
job_col = 'job_title'  # Update this
salary_col = 'salary'  # Update this

if job_col in df.columns and salary_col in df.columns:
    plt.figure(figsize=(12, 6))
    top_jobs = df.groupby(job_col)[salary_col].median().sort_values(ascending=False).head(15)
    sns.barplot(x=top_jobs.values, y=top_jobs.index, palette='coolwarm')
    plt.xlabel('Median Salary (INR)')
    plt.ylabel(job_col)
    plt.title('Median Salary by Job Title (Top 15)')
    plt.tight_layout()
    plt.savefig(os.path.join(RESULTS_DIR, 'figures', 'salary_by_job_title.png'), dpi=150, bbox_inches='tight')
    plt.show()
```

## Salary by Location

```python
# Salary by Location
location_col = 'location'  # Update this

if location_col in df.columns:
    plt.figure(figsize=(12, 6))
    top_locations = df.groupby(location_col)[salary_col].median().sort_values(ascending=False).head(15)
    sns.barplot(x=top_locations.values, y=top_locations.index, palette='viridis')
    plt.xlabel('Median Salary (INR)')
    plt.ylabel(location_col)
    plt.title('Median Salary by Location (Top 15)')
    plt.tight_layout()
    plt.savefig(os.path.join(RESULTS_DIR, 'figures', 'salary_by_location.png'), dpi=150, bbox_inches='tight')
    plt.show()
```

## Salary by Company

```python
# Top paying companies
company_col = 'company_name'  # Update this

if company_col in df.columns:
    plt.figure(figsize=(12, 6))
    top_companies = df.groupby(company_col)[salary_col].median().sort_values(ascending=False).head(15)
    sns.barplot(x=top_companies.values, y=top_companies.index, palette='magma')
    plt.xlabel('Median Salary (INR)')
    plt.ylabel(company_col)
    plt.title('Top 15 Highest Paying Companies')
    plt.tight_layout()
    plt.savefig(os.path.join(RESULTS_DIR, 'figures', 'top_paying_companies.png'), dpi=150, bbox_inches='tight')
    plt.show()
```

## Correlation Analysis

```python
# Correlation heatmap for numeric columns
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

if len(numeric_cols) > 1:
    plt.figure(figsize=(10, 8))
    corr_matrix = df[numeric_cols].corr()
    sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f')
    plt.title('Correlation Heatmap')
    plt.tight_layout()
    plt.savefig(os.path.join(RESULTS_DIR, 'figures', 'correlation_heatmap.png'), dpi=150, bbox_inches='tight')
    plt.show()
```

## Experience vs Salary

```python
# Experience vs Salary
exp_col = 'experience'  # Update this

if exp_col in df.columns:
    plt.figure(figsize=(10, 6))
    sns.boxplot(data=df, x=exp_col, y=salary_col, palette='Set2')
    plt.xlabel('Experience (Years)')
    plt.ylabel('Salary (INR)')
    plt.title('Salary Distribution by Experience')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig(os.path.join(RESULTS_DIR, 'figures', 'salary_by_experience.png'), dpi=150, bbox_inches='tight')
    plt.show()
```

## Save Analysis Results

```python
# Save aggregated statistics
salary_by_job = df.groupby(job_col)[salary_col].agg(['mean', 'median', 'min', 'max', 'count']).sort_values('median', ascending=False)
salary_by_job.to_csv(os.path.join(RESULTS_DIR, 'tables', 'salary_by_job_title.csv'))
print("Bivariate analysis results saved!")
```
